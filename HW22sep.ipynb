## Different Indexing Oprations On Name

```python
name = "Yash Pratapsing Rajput"
print(name)
print(len(name))
```

### First To Last (LHS)

```python
print(name[0], name[21])
```

### Last To First (RHS)

```python
print(name[-1], name[-22])
```

### Diff Ways

```python
print(name[0], name[16])
```
