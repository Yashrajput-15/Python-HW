```python
# Arithmatic oprator

s1 = 10        #int
s2 = 3.0       #float

print(s1+s2)     #Add
print(s1*s2)     #Multi
print(s1/s2)     #Div
print(s1//s2)    #FloorDiv
print(s1%s2)     #Modulo
```

```python
# Comparison oprator

print(s1>s2)
print(s1==10)
print(s2==3)
print(s1>=s2)
print(s1<=s2)
print(s1!=s2)
```

```python
# string

s3 = 'Yash'
s4 = 'Anna'

print(s3 +  "  "  + s4)
print(s3*4 + " " +s4*4)
print(s3 < s4)
print(s3 >= s4)
print(s3 <= 'A')
print(s3 != s4)
```

```python
# Boolean

s5 = True
s6 = False

print(s5 + s6)
print(s5 - s6)
print(s5 + s5)    # True + True = 1 + 1 =2
print(s6 + s6)    # False + False = 0 + 0 =0

print(s5 > s6)
print(s5 < s6)
print(s5 != s6)
```

```python
# Arrys

s7 = [2,4,6]
s8 = [1,3,5]

print(s7 + s8)
print(s7*3)
print(s7+ [9,8,7,9.0])

print(s7 > s8)
print(s7 >= s8)
print(s7 != s8)
print(s7 == s8)

```

```python
# if else & elif condition

score = int(input("Enter Your Score:"))

if score >= 80-90:
    print("Grade: A - Excellent")
elif score >= 60-70:
    print("Grade: B - Good")
else:
    print("Fail: Better Luck Next Time...!!!")


```
