##  Grade Calculator
### Create a variable score with a number between 0 and 100.
####  Conditions to check
1) Score 90 or above → Print "Grade: A"
2) Score 80 to 89 → Print "Grade: B"
3) Score 70 to 79 → Print "Grade: C"
4) Score 60 to 69 → Print "Grade: D"
5) Anything below 60 → Print "Grade: F"

```python
score = int(input("Enter Your Score: "))

if score >= 90:
    print("Grade: A" "" 'Excellant')
elif score >= 80:
    print("Grade: B" "" 'Very Good')
elif score >= 70:
    print("Grade: C" "" 'Good')
elif score >= 60:
    print("Grade: D" "" 'Average')
else:
    print("Grade: F" " " 'Better Luck...!')
```

## Positives, Negatives, and Zeros
### Goal: Create a variable named number and assign any integer to it.
### Conditions to check.
1) If the number is greater than 0, print "The number is positive."
2) If the number is less than 0, print "The number is negative."
3) Otherwise, print "The number is zero."

```python
num = int(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

## Traffic Light
### Goal: Create a variable Traffic signal and assign it a text value (like "red", "yellow", or "green").
### Conditions to check:
1) If the color is "red", print "Stop!"
2) If the color is "yellow", print "Slow down!"
3) If the color is "green", print "Go!"
4) If it is any other text, print "!"

```python
traffic_signal = "red"

if traffic_signal == "red":
    print("Stop!")
elif traffic_signal == "yellow":
    print("Slow down!")
elif traffic_signal == "green":
    print("Go!")
else:
    print("Signal broken...!")
```

## The Leap Year Checker
1) A year is a leap year if it is divisible by 4, but not divisible by 100 unless it is also divisible by 400.
2) Try using logical operators like and or or within your if statements.


```python
year = 2028

if year % 4 == 0 and year % 100 != 0 or year % 400 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```
